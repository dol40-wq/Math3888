```python
import PyDSTool as dst
import numpy as np
from matplotlib import pyplot as plt
from PyDSTool.Toolbox import phaseplane as pp
```

```python
DSargs = dst.args(name = 'ex')
pars = {
    'Cm': 1.0,
    'g_Na': 20.0,
    'g_K': 10.0,
    'g_l': 8.0,
    'E_Na': 60.0,
    'E_K': -90.0,
    'E_l': -80.0,
    'v1': -20.0,
    'k1': 15.0,
    'v3': -45.0,
    'k3': 5.0,
    'phi': 2.0,
    'I_app': 50.0
}
fnspecs = {
    'm_inf': (['V'], '1/(1+exp((v1-V)/k1))'),
    'n_inf': (['V'], '1/(1+exp((v3-V)/k3))'),
    'I_ion': (['V', 'n'], 'g_Na*m_inf(V)*(V-E_Na) + g_K*n*(V-E_K) + g_l*(V-E_l)'),
}
varspecs = {
    'V': '(-I_ion(V, n)+ I_app)/Cm',
    'n': 'phi*(n_inf(V)-n)',
}
ics = {
    'V': -65.0,
    'n': 1
}
domain = {
    'V': [-100, 100],
    'n': [0, 100]
}
DSargs.pars = pars
DSargs.varspecs = varspecs
DSargs.ics = ics
DSargs.fnspecs = fnspecs
DSargs.xdomain = domain
```

```python
ode_sys = dst.Generator.Vode_ODEsystem(DSargs)

# Define the range of time.
ode_sys.set(tdata=[0, 30])

print("Computing trajectory...")
trajectory = ode_sys.compute('dimensionless_run')
points = trajectory.sample()

fp_coord = pp.find_fixedpoints(ode_sys, n = 4, eps = 1e-8)[0]
print(fp_coord)
pycont = dst.ContClass(ode_sys)
pcargs = dst.args(name='EQ1', type='EP-C')
pcargs.freepars = ['I_app']
pcargs.MaxNumPoints = 500
pcargs.MaxStepSize = 1
pcargs.MinStepSize = 1e-3
pcargs.LocBifPoints = 'all'
pcargs.SaveEigen = True
pcargs.SaveJacobian = True

pcargs.initpoint = fp_coord
pcargs.verbosity = 2
pycont.newCurve(pcargs)
pycont["EQ1"].forward()
pycont["EQ1"].backward()
```

```python
# 4. Plot the Results
print("Plotting diagram.")

plt.figure(figsize=(10, 6))
pycont.display(('I_app', 'V'), stability=True)
plt.title("Bifurcation in (I_app, V) space")
```

```python
# This module is used to plot phaseplanes and nullclines
from PyDSTool.Toolbox import phaseplane as pp

# 2. Compute the trajectory

# We choose which integrator to use, in this case it uses scipy' VODE
ode_sys = dst.Generator.Vode_ODEsystem(DSargs)

# Define the range of time.
ode_sys.set(tdata=[0, 10])

print("Computing trajectory...")
trajectory = ode_sys.compute('dimensionless_run')
points = trajectory.sample()

# 3. Plotting Part A: Time Series and Basic Trajectory
print("Plotting trajectory part A...")
fig_a1, ax_a1 = plt.subplots(figsize = (8,8))
ax_a1.plot(points['t'], points['V'], 'b', lw =2, label = 'V(t)')
ax_a1.plot(points['t'], points['n'], 'g', lw = 2, label = "n(t)")
ax_a1.set_xlabel('Time')
ax_a1.set_ylabel("Voltage / Gating Variable")
ax_a1.set_title('Time series')
ax_a1.grid(True)
plt.legend()
plt.show()

# 4. Plot the phase trajectory
print("Plotting phase trajectory part A...")
fig_A2, ax_A2 = plt.subplots(figsize=(8, 8))
ax_A2.plot(points['V'], points['n'], 'k-', lw=2, label='Trajectory')
ax_A2.plot(ics['V'], ics['n'], 'go', ms=10, label=f'Start ({ics["V"]},{ics["n"]})')
ax_A2.set_title('Part A: Phase Trajectory', fontsize=16)
ax_A2.set_xlabel('V (Voltage)', fontsize=12)
ax_A2.set_ylabel('n (Gating Variable)', fontsize=12)
ax_A2.set_xlim(-0.1, 3)
ax_A2.set_ylim(-0.1, 3) 
plt.legend()
ax_A2.grid(True)
plt.show()

# 5. Plotting Part B: Adding the Vector Field
print("Plotting phase trajectory part B...")
fig_B, ax_B = plt.subplots(figsize=(8, 8))
ax_B.plot(points['V'], points['n'], 'k-', lw=2, label='Trajectory')
ax_B.plot(ics['V'], ics['n'], 'go', ms=10, label=f'Start ({ics["V"]},{ics["n"]})')

# Plot vector field, uses a scale exponent. Not a fan of the scale exponent, there is no option to normalise.
pp.plot_PP_vf(ode_sys, "V", "n", N=40, scale_exp=0)

ax_B.set_title('Part B: Phase Portrait with Vector Field', fontsize=16)
ax_B.set_xlabel('V (Voltage)', fontsize=12)
ax_B.set_ylabel('n (Gating Variable)', fontsize=12)
ax_B.set_title('Phase Trajectory', fontsize=16)
ax_B.set_xlabel('V', fontsize=12)
ax_B.set_ylabel('n', fontsize=12)
ax_B.set_xlim(-0.1, 3)
ax_B.set_ylim(-0.1, 3) 
ax_B.legend()
ax_B.set_aspect('equal', adjustable='box')
plt.show()


# 6 Plotting Part C: The Complete Phase Portrait

# First find the SS (fixed points). It returns a list as there can be multiple fixed points.

fp_coord = pp.find_fixedpoints(ode_sys, n=4, eps=1e-8)[0]

# Now find the nullcline, giving the fixed points will help with convergence.
nulls_s, nulls_c = pp.find_nullclines(ode_sys, 'V', 'n', n=3, eps=1e-8, max_step=0.05, fps = [fp_coord])






fig_C, ax_C = plt.subplots(figsize=(8, 8))

ax_C.plot(points['V'], points['n'], 'k-', lw=2, label='Trajectory')
ax_C.plot(ics['V'], ics['n'], 'go', ms=10, label=f'Start ({ics["V"]},{ics["n"]})')
ax_C.set_title('Part C: Complete Phase Portrait', fontsize=16)
ax_C.set_xlabel('V (Voltage)', fontsize=12)
ax_C.set_ylabel('n (Gating Variable)', fontsize=12)

# Plot vector field, uses a scale exponent. Not a fan of the scale exponent, there is no option to normalise. Preferably be done manually.
pp.plot_PP_vf(ode_sys, "V", "n", N=40, scale_exp=0)



# Plot the nullclines
plt.plot(nulls_s[:,0], nulls_s[:,1], 'r--', label='V-nullcline (dV/dt=0)')
plt.plot(nulls_c[:,0], nulls_c[:,1], 'b--', label='n-nullcline (dn/dt=0)')

# Plot the fixed point
ax_C.plot(fp_coord["V"], fp_coord["n"], 'm*', ms=15, label=f'Stable Steady State ({fp_coord["V"]:.2f},{fp_coord["n"]:.2f})')

ax_C.set_xlim(-0.1, 3)
ax_C.set_ylim(-0.1, 3) 
ax_C.legend()
ax_C.set_aspect('equal', adjustable='box')
plt.show()
```
